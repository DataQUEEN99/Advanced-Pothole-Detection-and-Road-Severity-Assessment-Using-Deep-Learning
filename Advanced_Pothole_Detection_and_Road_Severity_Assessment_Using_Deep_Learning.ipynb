```python
# ================================================================
# SMALL ADVANCED POTHOLE CNN PROJECT
# EfficientNetB0 + Augmentation + Evaluation + Severity + Risk
# GOOGLE COLAB - COMPLETE PROJECT IN ONE CELL
# ================================================================

# ----------------------------- IMPORTS -----------------------------

import os
import glob
import zipfile
import random
import shutil
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image

import tensorflow as tf

from tensorflow.keras import layers, Model
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.applications.efficientnet import preprocess_input
from tensorflow.keras.preprocessing.image import ImageDataGenerator

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    roc_curve
)

print("=" * 70)
print("ADVANCED POTHOLE CNN PROJECT")
print("=" * 70)

print("TensorFlow:", tf.__version__)


# ----------------------------- SETTINGS -----------------------------

SEED = 42
IMG_SIZE = 224
BATCH_SIZE = 32
EPOCHS = 10

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)


# ----------------------------- DATASET UPLOAD -----------------------------

print("\nUpload your pothole dataset ZIP.")
print("Expected structure:")
print("pothole_dataset/pothole/")
print("pothole_dataset/normal/")

from google.colab import files

uploaded = files.upload()

zip_files = [
    f for f in uploaded.keys()
    if f.lower().endswith(".zip")
]

if len(zip_files) == 0:
    raise ValueError(
        "Please upload a ZIP file containing pothole and normal folders."
    )

zip_path = zip_files[0]

extract_dir = "/content/pothole_data"

if os.path.exists(extract_dir):
    shutil.rmtree(extract_dir)

os.makedirs(extract_dir, exist_ok=True)

with zipfile.ZipFile(zip_path, "r") as z:
    z.extractall(extract_dir)

print("\nDataset extracted.")


# ----------------------------- FIND CLASS FOLDERS -----------------------------

def find_folder(root, names):

    for current_root, dirs, files_in_dir in os.walk(root):

        for d in dirs:

            if d.lower() in names:

                return os.path.join(current_root, d)

    return None


pothole_dir = find_folder(
    extract_dir,
    {"pothole", "potholes"}
)

normal_dir = find_folder(
    extract_dir,
    {
        "normal",
        "normal_road",
        "normal-road",
        "nopothole",
        "no_pothole"
    }
)

if pothole_dir is None or normal_dir is None:

    print("\nCould not automatically find folders.")

    print("Folders found:")

    for root, dirs, files_in_dir in os.walk(extract_dir):

        for d in dirs:
            print(os.path.join(root, d))

    raise ValueError(
        "\nDataset must contain folders named "
        "'pothole' and 'normal'."
    )

print("\nPothole folder:", pothole_dir)
print("Normal folder :", normal_dir)


# ----------------------------- FIND IMAGES -----------------------------

extensions = [
    "*.jpg",
    "*.jpeg",
    "*.png",
    "*.bmp",
    "*.webp"
]


def collect_images(folder):

    result = []

    for ext in extensions:

        result.extend(
            glob.glob(
                os.path.join(
                    folder,
                    "**",
                    ext
                ),
                recursive=True
            )
        )

    return sorted(
        list(set(result))
    )


pothole_images = collect_images(
    pothole_dir
)

normal_images = collect_images(
    normal_dir
)

print("\nDataset statistics")
print("-" * 50)

print(
    "Pothole images:",
    len(pothole_images)
)

print(
    "Normal images :",
    len(normal_images)
)

if len(pothole_images) < 5 or len(normal_images) < 5:

    raise ValueError(
        "Each class should contain at least 5 images."
    )


# ----------------------------- CREATE DATAFRAME -----------------------------

data = []

for path in pothole_images:

    data.append({
        "filepath": path,
        "label": "pothole"
    })

for path in normal_images:

    data.append({
        "filepath": path,
        "label": "normal"
    })

df = pd.DataFrame(data)

df = df.sample(
    frac=1,
    random_state=SEED
).reset_index(drop=True)

print("\nTotal images:", len(df))

print("\nClass distribution:")
print(df["label"].value_counts())


# ----------------------------- TRAIN / VALIDATION / TEST -----------------------------

from sklearn.model_selection import train_test_split

train_df, temp_df = train_test_split(
    df,
    test_size=0.30,
    stratify=df["label"],
    random_state=SEED
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["label"],
    random_state=SEED
)

print("\nDataset split")
print("-" * 50)

print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))


# ----------------------------- DATA AUGMENTATION -----------------------------

train_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_input,

    rotation_range=20,

    width_shift_range=0.12,

    height_shift_range=0.12,

    zoom_range=0.20,

    shear_range=0.10,

    horizontal_flip=True,

    brightness_range=(0.75, 1.25),

    fill_mode="nearest"
)

valid_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_input
)


train_generator = train_datagen.flow_from_dataframe(

    train_df,

    x_col="filepath",

    y_col="label",

    target_size=(IMG_SIZE, IMG_SIZE),

    batch_size=BATCH_SIZE,

    class_mode="binary",

    shuffle=True,

    seed=SEED
)


val_generator = valid_datagen.flow_from_dataframe(

    val_df,

    x_col="filepath",

    y_col="label",

    target_size=(IMG_SIZE, IMG_SIZE),

    batch_size=BATCH_SIZE,

    class_mode="binary",

    shuffle=False
)


test_generator = valid_datagen.flow_from_dataframe(

    test_df,

    x_col="filepath",

    y_col="label",

    target_size=(IMG_SIZE, IMG_SIZE),

    batch_size=BATCH_SIZE,

    class_mode="binary",

    shuffle=False
)


# ----------------------------- CLASS WEIGHTS -----------------------------

counts = train_df["label"].value_counts()

total = len(train_df)

class_weights = {

    train_generator.class_indices["normal"]:
        total / (2 * counts["normal"]),

    train_generator.class_indices["pothole"]:
        total / (2 * counts["pothole"])
}

print("\nClass indices:")
print(train_generator.class_indices)

print("\nClass weights:")
print(class_weights)


# ----------------------------- CNN MODEL -----------------------------

print("\n")
print("=" * 70)
print("BUILDING EFFICIENTNETB0")
print("=" * 70)

base_model = EfficientNetB0(

    include_top=False,

    weights="imagenet",

    input_shape=(
        IMG_SIZE,
        IMG_SIZE,
        3
    )
)

base_model.trainable = False


inputs = layers.Input(
    shape=(
        IMG_SIZE,
        IMG_SIZE,
        3
    )
)

x = base_model(
    inputs,
    training=False
)

x = layers.GlobalAveragePooling2D()(x)

x = layers.BatchNormalization()(x)

x = layers.Dropout(0.40)(x)

x = layers.Dense(
    128,
    activation="relu"
)(x)

x = layers.Dropout(0.25)(x)

outputs = layers.Dense(
    1,
    activation="sigmoid"
)(x)

model = Model(
    inputs,
    outputs
)


model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="binary_crossentropy",

    metrics=[
        "accuracy",

        tf.keras.metrics.Precision(
            name="precision"
        ),

        tf.keras.metrics.Recall(
            name="recall"
        ),

        tf.keras.metrics.AUC(
            name="auc"
        )
    ]
)


model.summary()


# ----------------------------- CALLBACKS -----------------------------

callbacks = [

    EarlyStopping(
        monitor="val_auc",
        patience=3,
        mode="max",
        restore_best_weights=True
    ),

    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.3,
        patience=2,
        min_lr=1e-7,
        verbose=1
    )
]


# ----------------------------- TRAIN -----------------------------

print("\n")
print("=" * 70)
print("TRAINING CNN")
print("=" * 70)

history = model.fit(

    train_generator,

    validation_data=val_generator,

    epochs=EPOCHS,

    class_weight=class_weights,

    callbacks=callbacks,

    verbose=1
)


# ----------------------------- OPTIONAL FINE TUNING -----------------------------

print("\n")
print("=" * 70)
print("FINE-TUNING")
print("=" * 70)

base_model.trainable = True

# Freeze most layers.
# Only the final layers are fine-tuned.

for layer in base_model.layers[:-30]:

    layer.trainable = False


model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-5
    ),

    loss="binary_crossentropy",

    metrics=[
        "accuracy",

        tf.keras.metrics.Precision(
            name="precision"
        ),

        tf.keras.metrics.Recall(
            name="recall"
        ),

        tf.keras.metrics.AUC(
            name="auc"
        )
    ]
)


fine_history = model.fit(

    train_generator,

    validation_data=val_generator,

    epochs=5,

    class_weight=class_weights,

    callbacks=callbacks,

    verbose=1
)


# ----------------------------- TEST PREDICTIONS -----------------------------

test_generator.reset()

probabilities = model.predict(
    test_generator,
    verbose=1
).ravel()

predictions = (
    probabilities >= 0.50
).astype(int)

true_labels = test_generator.classes


# ----------------------------- METRICS -----------------------------

accuracy = accuracy_score(
    true_labels,
    predictions
)

precision = precision_score(
    true_labels,
    predictions,
    zero_division=0
)

recall = recall_score(
    true_labels,
    predictions,
    zero_division=0
)

f1 = f1_score(
    true_labels,
    predictions,
    zero_division=0
)

auc = roc_auc_score(
    true_labels,
    probabilities
)


print("\n")
print("=" * 70)
print("FINAL MODEL PERFORMANCE")
print("=" * 70)

print(
    f"Accuracy  : {accuracy:.4f}"
)

print(
    f"Precision : {precision:.4f}"
)

print(
    f"Recall    : {recall:.4f}"
)

print(
    f"F1 Score  : {f1:.4f}"
)

print(
    f"ROC-AUC   : {auc:.4f}"
)


# ----------------------------- CLASSIFICATION REPORT -----------------------------

print("\nClassification Report")
print("-" * 70)

print(
    classification_report(
        true_labels,
        predictions,
        target_names=[
            "Normal Road",
            "Pothole"
        ],
        zero_division=0
    )
)


# ----------------------------- CONFUSION MATRIX -----------------------------

cm = confusion_matrix(
    true_labels,
    predictions
)

plt.figure(
    figsize=(7, 6)
)

plt.imshow(cm)

plt.title(
    "Pothole CNN — Confusion Matrix"
)

plt.xlabel(
    "Predicted Class"
)

plt.ylabel(
    "Actual Class"
)

plt.xticks(
    [0, 1],
    [
        "Normal",
        "Pothole"
    ]
)

plt.yticks(
    [0, 1],
    [
        "Normal",
        "Pothole"
    ]
)

for i in range(2):

    for j in range(2):

        plt.text(
            j,
            i,
            cm[i, j],
            ha="center",
            va="center",
            fontsize=18
        )

plt.colorbar()

plt.tight_layout()

plt.show()


# ----------------------------- TRAINING CURVES -----------------------------

train_acc = (
    history.history["accuracy"]
    +
    fine_history.history["accuracy"]
)

val_acc = (
    history.history["val_accuracy"]
    +
    fine_history.history["val_accuracy"]
)

train_loss = (
    history.history["loss"]
    +
    fine_history.history["loss"]
)

val_loss = (
    history.history["val_loss"]
    +
    fine_history.history["val_loss"]
)


plt.figure(
    figsize=(10, 5)
)

plt.plot(
    train_acc,
    label="Training Accuracy"
)

plt.plot(
    val_acc,
    label="Validation Accuracy"
)

plt.title(
    "Training vs Validation Accuracy"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.legend()

plt.grid(
    alpha=0.25
)

plt.show()


plt.figure(
    figsize=(10, 5)
)

plt.plot(
    train_loss,
    label="Training Loss"
)

plt.plot(
    val_loss,
    label="Validation Loss"
)

plt.title(
    "Training vs Validation Loss"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Loss"
)

plt.legend()

plt.grid(
    alpha=0.25
)

plt.show()


# ----------------------------- ROC CURVE -----------------------------

fpr, tpr, thresholds = roc_curve(
    true_labels,
    probabilities
)

plt.figure(
    figsize=(8, 6)
)

plt.plot(
    fpr,
    tpr,
    label=f"ROC-AUC = {auc:.3f}"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "ROC Curve"
)

plt.legend()

plt.grid(
    alpha=0.25
)

plt.show()


# ----------------------------- CONFIDENCE DISTRIBUTION -----------------------------

plt.figure(
    figsize=(10, 6)
)

plt.hist(
    probabilities[
        true_labels == 1
    ],
    bins=20,
    alpha=0.65,
    label="Actual Potholes"
)

plt.hist(
    probabilities[
        true_labels == 0
    ],
    bins=20,
    alpha=0.65,
    label="Actual Normal Roads"
)

plt.axvline(
    0.50,
    linestyle="--",
    label="Decision Threshold"
)

plt.xlabel(
    "Pothole Probability"
)

plt.ylabel(
    "Number of Images"
)

plt.title(
    "Model Confidence Distribution"
)

plt.legend()

plt.grid(
    alpha=0.25
)

plt.show()


# ----------------------------- VISUAL SEVERITY -----------------------------

def calculate_visual_severity(image):

    arr = np.asarray(
        image.convert("RGB")
    ).astype(
        np.float32
    ) / 255.0

    gray = (
        0.299 * arr[:, :, 0]
        +
        0.587 * arr[:, :, 1]
        +
        0.114 * arr[:, :, 2]
    )

    dark_ratio = np.mean(
        gray < 0.30
    )

    contrast = np.std(
        gray
    )

    horizontal_edges = np.abs(
        np.diff(
            gray,
            axis=1
        )
    )

    vertical_edges = np.abs(
        np.diff(
            gray,
            axis=0
        )
    )

    edge_density = (
        np.mean(
            horizontal_edges > 0.10
        )
        +
        np.mean(
            vertical_edges > 0.10
        )
    ) / 2

    raw_score = (
        0.40 * dark_ratio
        +
        0.30 * contrast
        +
        0.30 * edge_density
    )

    score = np.clip(
        raw_score * 160,
        0,
        100
    )

    if score < 30:

        severity = "LOW"

    elif score < 60:

        severity = "MODERATE"

    elif score < 80:

        severity = "HIGH"

    else:

        severity = "CRITICAL"

    return float(score), severity


# ----------------------------- SINGLE IMAGE ANALYZER -----------------------------

def analyze_pothole_image(
    image_path,
    show=True
):

    image = Image.open(
        image_path
    ).convert(
        "RGB"
    )

    resized = image.resize(
        (
            IMG_SIZE,
            IMG_SIZE
        )
    )

    arr = np.asarray(
        resized
    ).astype(
        np.float32
    )

    arr = np.expand_dims(
        arr,
        axis=0
    )

    arr = preprocess_input(
        arr
    )

    pothole_probability = float(
        model.predict(
            arr,
            verbose=0
        )[0][0]
    )

    if pothole_probability >= 0.50:

        prediction = "POTHOLE"

        confidence = pothole_probability

    else:

        prediction = "NORMAL ROAD"

        confidence = 1 - pothole_probability


    severity_score, severity = (
        calculate_visual_severity(
            image
        )
    )


    if prediction == "POTHOLE":

        risk_score = (
            0.65 * confidence * 100
            +
            0.35 * severity_score
        )

    else:

        risk_score = (
            0.20 * confidence * 100
        )


    risk_score = float(
        np.clip(
            risk_score,
            0,
            100
        )
    )


    if risk_score < 30:

        risk = "LOW"

    elif risk_score < 60:

        risk = "MODERATE"

    elif risk_score < 80:

        risk = "HIGH"

    else:

        risk = "CRITICAL"


    if show:

        plt.figure(
            figsize=(9, 7)
        )

        plt.imshow(
            image
        )

        plt.axis(
            "off"
        )

        plt.title(
            f"{prediction}\n"
            f"Confidence: {confidence:.1%}\n"
            f"Severity: {severity} "
            f"({severity_score:.1f}/100)\n"
            f"Road Risk: {risk} "
            f"({risk_score:.1f}/100)",
            fontsize=14
        )

        plt.show()


        print("\n")
        print("=" * 60)
        print("POTHOLE ANALYSIS")
        print("=" * 60)

        print(
            "Prediction      :",
            prediction
        )

        print(
            "Confidence      :",
            f"{confidence:.2%}"
        )

        print(
            "Severity Score  :",
            f"{severity_score:.1f}/100"
        )

        print(
            "Severity        :",
            severity
        )

        print(
            "Risk Score      :",
            f"{risk_score:.1f}/100"
        )

        print(
            "Risk Level      :",
            risk
        )


    return {
        "prediction": prediction,
        "confidence": confidence,
        "severity_score": severity_score,
        "severity": severity,
        "risk_score": risk_score,
        "risk": risk
    }


# ----------------------------- SAMPLE TEST IMAGES -----------------------------

print("\n")
print("=" * 70)
print("SAMPLE TEST PREDICTIONS")
print("=" * 70)

sample_paths = test_df.sample(
    min(6, len(test_df)),
    random_state=SEED
)["filepath"].tolist()


plt.figure(
    figsize=(15, 10)
)

for i, path in enumerate(sample_paths):

    image = Image.open(
        path
    ).convert(
        "RGB"
    )

    resized = image.resize(
        (
            IMG_SIZE,
            IMG_SIZE
        )
    )

    arr = np.asarray(
        resized
    ).astype(
        np.float32
    )

    arr = np.expand_dims(
        arr,
        axis=0
    )

    arr = preprocess_input(
        arr
    )

    probability = float(
        model.predict(
            arr,
            verbose=0
        )[0][0]
    )

    prediction = (
        "POTHOLE"
        if probability >= 0.50
        else "NORMAL"
    )

    actual = (
        "POTHOLE"
        if path in pothole_images
        else "NORMAL"
    )

    plt.subplot(
        2,
        3,
        i + 1
    )

    plt.imshow(
        image
    )

    plt.axis(
        "off"
    )

    plt.title(
        f"Actual: {actual}\n"
        f"Predicted: {prediction}\n"
        f"Pothole probability: {probability:.2f}"
    )

plt.tight_layout()

plt.show()


# ----------------------------- SAVE MODEL -----------------------------

model_path = (
    "/content/"
    "advanced_pothole_cnn.keras"
)

model.save(
    model_path
)

print(
    "\nModel saved to:",
    model_path
)


# ----------------------------- FINAL REPORT -----------------------------

print("\n")
print("=" * 70)
print("FINAL PROJECT SUMMARY")
print("=" * 70)

print(
    f"""
Project:
Advanced Pothole Detection using CNN

Dataset
-------
Total images       : {len(df)}
Training images    : {len(train_df)}
Validation images  : {len(val_df)}
Testing images     : {len(test_df)}

Architecture
------------
Backbone           : EfficientNetB0
Transfer Learning  : Yes
Fine Tuning        : Yes
Data Augmentation  : Yes
Class Balancing    : Yes
Dropout            : Yes
Batch Normalization: Yes

Performance
-----------
Accuracy           : {accuracy:.2%}
Precision          : {precision:.2%}
Recall             : {recall:.2%}
F1 Score           : {f1:.2%}
ROC-AUC            : {auc:.2%}

Additional Features
-------------------
Confidence analysis
Visual severity score
Road-risk score
Confusion matrix
ROC curve
Training curves
Error analysis
Sample predictions

Saved model:
{model_path}

IMPORTANT:
The severity/risk score is a visual estimate.
It does not represent actual pothole depth.
Physical depth requires depth/stereo/LiDAR/calibrated
measurement data.
"""
)

print("=" * 70)
print("PROJECT COMPLETE")
print("=" * 70)


# ----------------------------- HOW TO TEST YOUR OWN IMAGE -----------------------------

print("""
===============================================================
TEST YOUR OWN ROAD IMAGE
===============================================================

Run:

from google.colab import files

uploaded = files.upload()

Then:

analyze_pothole_image(
    "/content/YOUR_IMAGE_NAME.jpg"
)

===============================================================
""")
```
